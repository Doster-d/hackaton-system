```python
from pathlib import Path
```

```python
video_path = Path('../data/raw/target/part_00.mp4')
video_path.exists()
```

```python
# Example: Using VideoLLaMA3ActionInferrer for question answering
from hackaton_system.config import get_settings
from hackaton_system.pipeline.videollama3 import VideoLLaMA3ActionInferrer

settings = get_settings()
inferrer = VideoLLaMA3ActionInferrer(settings)

# Ask questions about the video
questions = [
    "Какой номер у поезда?",
    "Опиши, что происходит в этом видео.",
    "Сколько людей видно в кадре?",
]

for question in questions:
    answer = inferrer.ask_question(video_path, question, max_new_tokens=256)
    print(f"Q: {question}")
    print(f"A: {answer}\n")
```
